```python
import time
from functools import wraps
```

```python
def timeit(fn):
    @wraps(fn)
    def inner(*args, **kwargs):
        start = time.time()
        try:
            return fn(*args, **kwargs)
        finally:
            print(f"time of {fn.__name__}: {time.time() - start}")
    return inner
```

```python
@timeit
def add(a, b):
    time.sleep(a)
    return a + b
```

```python
add(0.1, 0.9)
```

```python
@timeit
def check_data(a):
    time.sleep(a)
    raise ValueError("wrong")
```

```python
def timeit(fn):
    count = 0
    sum_time = 0
    
    @wraps(fn)
    def inner(*args, **kwargs):
        nonlocal count, sum_time
        
        start = time.time()
        try:
            return fn(*args, **kwargs)
        finally:
            count += 1
            sum_time += time.time() - start
            print(f"avg time of {fn.__name__}: {sum_time / count}")
    return inner
```

```python
@timeit
def add(a, b):
    time.sleep(a)
    return a + b
```

```python
add(1, 2)
add(0.8, 1)
add(0.3, 1)
```

```python
from collections import deque
```

```python
def timeit_k_last(k):
    def timeit(fn):
        times = deque()
        count = 0
        sum_time = 0
        
        @wraps(fn)
        def inner(*args, **kwargs):
            nonlocal count, sum_time
            
            start = time.time()
            try:
                return fn(*args, **kwargs)
            finally:
                sum_time += time.time() - start
                times.append(time.time() - start)

                if count >= k:
                    sum_time -= times.popleft()
                else:
                    count += 1

                print(f"avg time of {k=} calls {fn.__name__}: {sum_time / count}")
        return inner
    return timeit
```

```python
@timeit_k_last(2)
def add(a, b):
    time.sleep(a)
    return a + b
```

```python
add(1, 2)
add(0.8, 1)
add(0.3, 1)
```
