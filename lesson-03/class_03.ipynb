```python
def timeit_k_last(k):
    def timeit(fn):
        times = deque()
        count = 0
        sum_time = 0
        
        @wraps(fn)
        def inner(*args, **kwargs):
            nonlocal count, sum_time

            start = time.time()
            try:
                return fn(*args, **kwargs)
            finally:
                sum_time += time.time() - start
                times.append(time.time() - start)

                if count >= k:
                    sum_time -= times.popleft()
                else:
                    count += 1

                print(f"avg time of {k=} calls {fn.__name__}: {sum_time / count}")
        return inner
    return timeit
```

```python
class ClassAttr:
    name = "cls_name"
    __cls_private = "cls_private"

    __custom_attr__ = "data"  # not recommended!

    def __init__(self, val):
        self.val = val
        self._protected = "protected"
        self.__private = "private"

    def print(self):
        print(
            f"{self.val=}, {self._protected=}, {self.__private=}, "
            f"{self.name=}, {self.__cls_private=}"
        )

    def print_new(self):
        print(f"{self.__private_3=}")

    @classmethod
    def update_name(cls, new_name):
        print(f"update_name {cls=}, {new_name=}")
        cls.name = new_name

    @staticmethod
    def process_data(val):
        print(f"process_data {val=}")


item = ClassAttr(10)
```

```python
class Num:
    def __init__(self, n):
        self.n = n

    def __add__(self, other):
        print(f"__add__ {other=}")
        return self.n + other

    def __radd__(self, other):
        print(f"__radd__ {other=}")
        return self.n + other


n = Num(99)

print(n + 10)
print(10 + n) # 10.__add__(n) -> err -> n.__radd__(10) -> 109
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age

    @property
    def age(self):
        return self.__age


class Student(Person):
    def format_data(self):
        return f"{self=}, {self.name=}, {self.age=}"


steve = Student("Steve", 99)
steve.format_data()
```
