```python
class DataProcessor(threading.Thread):
    def __init__(self, n):
        super().__init__()
        self.n = n
        self.result = 0

    def run(self):
        print(f"started {self.name=}, {self.ident=}, {self.daemon=}")

        for i in range(self.n):
            print(f"run: {i=}, {self.result=}")
            self.result += i
            time.sleep(1)
        
        print(f"finished {self.name=}, {self.ident=}, {self.daemon=}")


data = DataProcessor(10)
print(f"before start {data.result=}")

data.start()

print(f"after  start {data.result=}")
```

## Состояние гонки

```python
%%time


N = 10 ** 8
N_THREADS = 10
N_JOBS = N // N_THREADS


class PyObject:
    def __init__(self):
        self.ref_counter = 0


obj = PyObject()
print(f"start {obj.ref_counter=}")


def counter(a, b):
    print(f"counter[{a}, {b}], name={threading.current_thread().name} -- started")
    
    while a < b:
        a += 1
        tmp = obj.ref_counter
        x = min(tmp, 10)
        tmp += 1
        obj.ref_counter = tmp

    print(f"counter[{a}, {b}], name={threading.current_thread().name} -- finished")


threads = [
    threading.Thread(
        target=counter,
        args=(i * N_JOBS, (i + 1) * N_JOBS),
        name=f"th_counter_{i}",
    )
    for i in range(N_THREADS)
]

for th in threads:
    th.start()

for th in threads:
    th.join()


print(f"finish {obj.ref_counter=}, {obj.ref_counter == N=}")
```

```python
%%time


N = 10 ** 8
N_THREADS = 10
N_JOBS = N // N_THREADS


class PyObject:
    def __init__(self):
        self.ref_counter = 0


obj = PyObject()
print(f"start {obj.ref_counter=}")

lock = threading.Lock()


def counter(a, b):
    print(f"counter[{a}, {b}], name={threading.current_thread().name} -- started")

    while a < b:
        a += 1

        with lock:
            tmp = obj.ref_counter
            x = min(tmp, 10)
            tmp += 1
            obj.ref_counter = tmp

    print(f"counter[{a}, {b}], name={threading.current_thread().name} -- finished")


threads = [
    threading.Thread(
        target=counter,
        args=(i * N_JOBS, (i + 1) * N_JOBS),
        name=f"th_counter_{i}",
    )
    for i in range(N_THREADS)
]

for th in threads:
    th.start()

for th in threads:
    th.join()


print(f"finish {obj.ref_counter=}, {obj.ref_counter == N=}")
```

```python
%%time


N = 20
N_THREADS = 10
N_JOBS = N // N_THREADS

URL = "https://www.python.org/"
URLS = [URL] * N


def fetch_urls(urls, sem):
    print(f"fetch_urls, name={threading.current_thread().name} -- started")
    t1 = time.time()
    for url in urls:
        #time.sleep(1)
        with sem:
            resp = requests.get(url)

    t2 = time.time()
    print(f"fetch_urls, name={threading.current_thread().name} -- finished", t2 - t1)


def run():
    sem = threading.Semaphore(10)

    threads = [
        threading.Thread(
            target=fetch_urls,
            args=(URLS[i * N_JOBS : (i + 1) * N_JOBS], sem),
            name=f"th_counter_{i}",
        )
        for i in range(N_THREADS)
    ]
    
    for th in threads:
        th.start()
    
    for th in threads:
        th.join()

run()
```

```python
%%time


N = 20
N_THREADS = 10
N_JOBS = N // N_THREADS

URL = "https://www.python.org/"
URLS = [URL] * N


def fetch_urls(que):
    print(f"fetch_urls, name={threading.current_thread().name} -- started")
    t1 = time.time()
    cnt = 0

    while True:
        url = que.get()
        if url is None:
            que.put(url)
            break

        resp = requests.get(url)
        cnt += 1

    t2 = time.time()
    print(f"fetch_urls, name={threading.current_thread().name} -- finished", t2 - t1, cnt)


def run():
    que = queue.Queue()
    for url in URLS:
        que.put(url)
    que.put(None)

    threads = [
        threading.Thread(
            target=fetch_urls,
            args=(que,),
            name=f"th_counter_{i}",
        )
        for i in range(N_THREADS)
    ]
    
    for th in threads:
        th.start()
    
    for th in threads:
        th.join()

run()
```
