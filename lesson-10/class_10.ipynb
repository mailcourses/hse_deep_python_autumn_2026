```python
class Student:
    def __init__(self, name, monitor=None):
        self.name = name
        self.monitor = monitor

    def __str__(self):
        return f"Student[name={self.name}, id={id(self)}]"

    def __del__(self):
        print(f"__del__ {str(self)}")
```

```python
steve = Student("steve")
```

```python
sys.getrefcount(steve)
```

```python
del steve
```

```python
steve = Student("steve")
print(sys.getrefcount(steve))
del steve
```

```python
steve = Student("steve")
tom = Student("tom", monitor=steve)
steve.monitor = tom

print(f"{sys.getrefcount(steve)=}, {sys.getrefcount(tom)=}")

del steve, tom
```

```python
import ctypes
```

```python
class PyObject(ctypes.Structure):
    _fields_ = [("refcnt", ctypes.c_long)]


def get_refcnt(obj_id):
    return PyObject.from_address(obj_id).refcnt


def get_refcnt2(obj_id):
    return ctypes.c_long.from_address(obj_id).value
```

```python
steve = Student("steve")
tom = Student("tom", monitor=steve)
steve.monitor = tom

steve_id = id(steve)
tom_id = id(tom)

print(f"111: {sys.getrefcount(steve)=}, {sys.getrefcount(tom)=}")
print(f"222: {get_refcnt(steve_id)=}, {get_refcnt(tom_id)=}, {get_refcnt2(steve_id)=}, {get_refcnt2(tom_id)=}")

del steve, tom

print(f"333: {get_refcnt(steve_id)=}, {get_refcnt(tom_id)=}, {get_refcnt2(steve_id)=}, {get_refcnt2(tom_id)=}")

gc.collect()

print(f"444: {get_refcnt(steve_id)=}, {get_refcnt(tom_id)=}, {get_refcnt2(steve_id)=}, {get_refcnt2(tom_id)=}")
```
